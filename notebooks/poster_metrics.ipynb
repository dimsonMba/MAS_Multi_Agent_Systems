# MAS Resilience Poster Metrics

This notebook visualizes metrics exported by `run.py` into `data/results/`.

It focuses on:
- alive vs failed agents over time
- operational ratio trend
- safety hazard timeline
- summary numbers for poster captions

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")

results_dir = Path("../data/results")
csv_files = sorted(results_dir.glob("metrics_*.csv"), key=lambda p: p.stat().st_mtime)

if not csv_files:
    raise FileNotFoundError("No metrics CSV found. Run `python run.py` first.")

latest_csv = csv_files[-1]
print(f"Using file: {latest_csv}")
df = pd.read_csv(latest_csv)
df.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(df["step"], df["alive_agents"], label="Alive Agents", linewidth=2)
ax.plot(df["step"], df["failed_agents"], label="Failed Agents", linewidth=2)
ax.set_title("Alive vs Failed Agents Over Time")
ax.set_xlabel("Step")
ax.set_ylabel("Agent Count")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(df["step"], df["operational_ratio"], color="tab:blue", linewidth=2)
ax.set_ylim(0, 1.05)
ax.set_title("Operational Ratio (Resilience)")
ax.set_xlabel("Step")
ax.set_ylabel("Operational Ratio")
plt.tight_layout()
plt.show()
```

```python
hazard_numeric = df["hazard_flag"].astype(int)

fig, ax = plt.subplots(figsize=(9, 2.8))
ax.step(df["step"], hazard_numeric, where="post", color="tab:red", linewidth=2)
ax.set_title("Safety Hazard Timeline")
ax.set_xlabel("Step")
ax.set_ylabel("Hazard")
ax.set_yticks([0, 1])
ax.set_yticklabels(["No", "Yes"])
plt.tight_layout()
plt.show()
```

```python
summary = {
    "total_steps": int(df["step"].max()),
    "final_alive_agents": int(df["alive_agents"].iloc[-1]),
    "final_failed_agents": int(df["failed_agents"].iloc[-1]),
    "total_failure_events": int(df["total_failure_events"].iloc[-1]),
    "mean_operational_ratio": float(df["operational_ratio"].mean()),
    "hazard_steps": int(df["hazard_flag"].astype(int).sum()),
}

pd.Series(summary, name="poster_summary")
```
